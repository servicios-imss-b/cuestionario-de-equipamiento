```python
# Left join: agregar nombre_de_la_unidad de catalogo_clues a clues
import pandas as pd

catalogo_nombres = (
    catalogo_clues[['clues_imb', 'nombre_de_la_unidad']]
    .assign(_clues_key=lambda df: df['clues_imb'].astype('string').str.strip().str.upper())
    .drop_duplicates('_clues_key', keep='last')
    [['_clues_key', 'nombre_de_la_unidad']]
)

clues = (
    clues
    .drop(columns=['nombre_de_la_unidad'], errors='ignore')
    .assign(_clues_key=lambda df: df['CLUES'].astype('string').str.strip().str.upper())
    .merge(catalogo_nombres, on='_clues_key', how='left', validate='many_to_one')
    .drop(columns=['_clues_key'])
)

print(f'Filas en clues: {len(clues):,}')
print(f'Filas con nombre_de_la_unidad: {clues["nombre_de_la_unidad"].notna().sum():,}')
print(f'Filas sin coincidencia: {clues["nombre_de_la_unidad"].isna().sum():,}')
clues.head()
```

```python
# Actualizar los nombres de unidad en src/data/units.json sin modificar los demás campos
import json
from pathlib import Path

units_json_path = Path(fr"C:\Users\{usuario}\Downloads\cuestionario-de-equipamiento\cuestionario-de-equipamiento\src\data\units.json")

with units_json_path.open('r', encoding='utf-8') as archivo:
    units_json = json.load(archivo)

nombre_por_clues = (
    clues[['CLUES', 'nombre_de_la_unidad']]
    .dropna(subset=['CLUES', 'nombre_de_la_unidad'])
    .assign(CLUES=lambda df: df['CLUES'].astype('string').str.strip().str.upper())
    .drop_duplicates('CLUES', keep='last')
    .set_index('CLUES')['nombre_de_la_unidad']
    .to_dict()
)

actualizadas = 0
for unidad in units_json:
    clave = str(unidad.get('clues', '')).strip().upper()
    if clave in nombre_por_clues:
        nuevo_nombre = str(nombre_por_clues[clave]).strip()
        if unidad.get('name') != nuevo_nombre:
            unidad['name'] = nuevo_nombre
            actualizadas += 1

with units_json_path.open('w', encoding='utf-8') as archivo:
    json.dump(units_json, archivo, ensure_ascii=False, indent=2)
    archivo.write('\n')

print(f'Registros en JSON: {len(units_json):,}')
print(f'Nombres actualizados: {actualizadas:,}')
print(f'Archivo actualizado: {units_json_path}')
```
